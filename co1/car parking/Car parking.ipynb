```python
import datetime
import math

class ParkingSystem:
    def __init__(self, total_slots=100, rate_per_hour=20):
        self.total_slots = total_slots
        self.rate_per_hour = rate_per_hour
        self.available_slots = set(range(1, total_slots + 1))
        self.parking_slots = {}
        self.active_vehicles = {}

    def display_availability(self):
        total_available = len(self.available_slots)
        print(f"\n[STATUS] Available Slots: {total_available} / {self.total_slots}")
        if total_available > 0:
            sorted_slots = sorted(list(self.available_slots))
            display_count = min(10, total_available)
            print(f"Next vacant slots: {sorted_slots[:display_count]}" + (" ..." if total_available > 10 else ""))
        else:
            print("[ALERT] Parking Area is completely FULL!")

    def allocate_slot(self, vehicle_number):
        if not self.available_slots:
            print("[ERROR] Entry Denied: Parking area is full.")
            return None
        
        if vehicle_number in self.active_vehicles:
            assigned_slot = self.active_vehicles[vehicle_number]
            print(f"[WARNING] Vehicle {vehicle_number} is already inside at Slot #{assigned_slot}.")
            return None

        # Allocate the lowest vacant slot number
        slot = min(self.available_slots)
        self.available_slots.remove(slot)
        
        self.parking_slots[slot] = {
            "vehicle_number": vehicle_number,
            "entry_time": datetime.datetime.now()
        }
        self.active_vehicles[vehicle_number] = slot
        
        print(f"[SUCCESS] Vehicle {vehicle_number} successfully parked in Slot #{slot}.")
        return slot

    def release_slot(self, vehicle_number):
        if vehicle_number not in self.active_vehicles:
            print(f"[ERROR] Vehicle {vehicle_number} was not found in the active parking directory.")
            return None

        slot = self.active_vehicles.pop(vehicle_number)
        entry_time = self.parking_slots[slot]["entry_time"]
        exit_time = datetime.datetime.now()
        
        # Calculate elapsed hours rounded up
        duration = exit_time - entry_time
        hours_parked = max(1, math.ceil(duration.total_seconds() / 3600))
        total_fee = hours_parked * self.rate_per_hour

        print(f"[SUCCESS] Vehicle {vehicle_number} vacated Slot #{slot}.")
        print(f"  -> Duration: {hours_parked} hour(s)")
        print(f"  -> Calculated Fee: ${total_fee} (Rate: ${self.rate_per_hour}/hr)")

        # Clear tracking variables
        self.parking_slots.pop(slot)
        self.available_slots.add(slot)
        
        return total_fee

def main():
    system = ParkingSystem(total_slots=100, rate_per_hour=20)
    
    while True:
        print("\n====================================")
        print("    PARKING MANAGEMENT CONSOLE      ")
        print("====================================")
        print("1. View Current Slot Availability")
        print("2. Park Incoming Vehicle (Check-In)")
        print("3. Process Departing Vehicle (Check-Out)")
        print("4. Exit Management Program")
        
        choice = input("Select an option (1-4): ").strip()

        if choice == "1":
            system.display_availability()
        elif choice == "2":
            vnum = input("Enter vehicle license number: ").strip().upper()
            if vnum:
                system.allocate_slot(vnum)
            else:
                print("[ERROR] Invalid input. Vehicle license number cannot be blank.")
        elif choice == "3":
            vnum = input("Enter vehicle license number to exit: ").strip().upper()
            if vnum:
                system.release_slot(vnum)
            else:
                print("[ERROR] Invalid input. Vehicle license number cannot be blank.")
        elif choice == "4":
            print("\nShutting down console system. Goodbye!")
            break
        else:
            print("[ERROR] Invalid choice. Please enter a valid number (1-4).")

if __name__ == "__main__":
    main()
```
