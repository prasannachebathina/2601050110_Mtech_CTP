```python
def get_student_data():
    students = {}
    n = int(input("Enter number of students: "))

    for i in range(n):
        print(f"\nEnter details for Student {i + 1}")
        name = input("Enter name: ")
        total_classes = int(input("Enter total working classes: "))
        attended = int(input("Enter attended classes: "))

        students[name] = {
            "total": total_classes,
            "attended": attended
        }

    return students


def calculate_percentage(students):
    percentages = {}

    for name, data in students.items():
        percentage = (data["attended"] / data["total"]) * 100
        percentages[name] = round(percentage, 2)

    return percentages


def get_status(percentage):
    if percentage >= 75:
        return "Eligible"
    else:
        return "Shortage"


def display_report(students, percentages):
    print("\n========== ATTENDANCE REPORT ==========")

    for name in students:
        total = students[name]["total"]
        attended = students[name]["attended"]
        percentage = percentages[name]
        status = get_status(percentage)

        print("\nStudent Name:", name)
        print("Total Classes:", total)
        print("Classes Attended:", attended)
        print("Attendance Percentage:", percentage, "%")
        print("Status:", status)

    highest = max(percentages, key=percentages.get)
    lowest = min(percentages, key=percentages.get)

    average = sum(percentages.values()) / len(percentages)

    print("\n========== CLASS SUMMARY ==========")
    print("Highest Attendance:", highest, "-", percentages[highest], "%")
    print("Lowest Attendance:", lowest, "-", percentages[lowest], "%")
    print("Class Average:", round(average, 2), "%")


def main():
    students = get_student_data()
    percentages = calculate_percentage(students)
    display_report(students, percentages)


if __name__ == "__main__":
    main()
```
